Problem 5 -Password strength check


```python
password = "lusha@2003"

length_ok = len(password) >= 8
contains_at = "@" in password
strong = length_ok and contains_at

print(f"Password: {password}")
print(f"Length OK (>=8): {length_ok}")
print(f"Contains '@': {contains_at}")
print(f"Strong: {strong}")
```

Problem 9- sort products by price

```python
products = [("Notebook", 250),
          ("Pen", 20),
          ("Backpack", 1800),
          ("Eraser", 15),
          ("Water Bottle", 350)]
sorted_products = sorted(products, key=lambda x: x[1])
print(sorted_products)
```

Problem 10 - Filter the passing students

```python
scores = {"Anika": 35,
          "Rohan": 72,
          "Meera": 40,
          "Kabir": 28,
          "Diya": 88,
          "Ishaan": 55}

passing_students = list(map(lambda x: x[0],

                        filter(lambda x: x[1] >= 40, scores.items())))

print("Passing students:", passing_students)
```
