# MP2: Personal Expense Tracker
**Track Chosen:** Track A - Personal Expense Tracker

### Description
A menu-driven tool that tracks daily expenses by category (Food, Transport, etc.) and calculates total expenditure.
It preserves transaction history between sessions by storing data in a local JSON file.

### How to Run
1. Run all code cells in sequential order.
2. Call `main()` in the final cell to launch the program interface.


```python
import json
import os

DATA_FILE = "expenses.json"
```

```python
def load_expenses(filename=DATA_FILE):
    """Loads expenses from a JSON file. Returns an empty list if file doesn't exist."""
    if not os.path.exists(filename):
        return []
    try:
        with open(filename, "r") as f:
            return json.load(f)
    except (json.JSONDecodeError, IOError):
        print("Warning: Could not read data file. Starting with an empty list.")
        return []

def save_expenses(expenses, filename=DATA_FILE):
    """Saves the expenses list to a JSON file."""
    try:
        with open(filename, "w") as f:
            json.dump(expenses, f, indent=4)
        print("Data saved successfully.")
    except IOError as e:
        print(f"Error saving data: {e}")
```

```python
def add_expense(expenses):
    """Prompts user for input and adds a validated expense record."""
    category = input("Category (e.g., Food, Transport): ").strip().capitalize()
    if not category:
        category = "Uncategorized"
# Robust input validation handling bad numerical input
    while True:
        amount_str = input("Amount: ").strip()
        try:
            amount = float(amount_str)
            if amount < 0:
                print("Amount cannot be negative. Try again.")
                continue
            break
        except ValueError:
            print("Invalid input! Please enter a valid number for amount.")

    description = input("Description: ").strip()

    expense = {
        "category": category,
        "amount": amount,
        "description": description
    }
    expenses.append(expense)
    print("Added successfully.")

def view_all_expenses(expenses):
    """Prints all recorded expenses in a clean tabular format."""
    if not expenses:
        print("\nNo expenses recorded yet.")
        return

    print("\n=== All Expenses ===")
    for idx, item in enumerate(expenses, 1):
        print(f"{idx}. [{item['category']}] Rs {item['amount']:.2f} - {item['description']}")

def view_totals_by_category(expenses):
    """Calculates and displays summary metrics by category and overall total."""
    if not expenses:
        print("\nNo expenses recorded yet.")
        return

    totals = {}
    grand_total = 0.0

    for item in expenses:
        cat = item['category']
        amt = item['amount']
        totals[cat] = totals.get(cat, 0.0) + amt
        grand_total += amt

    print("\n=== Expenses Summary ===")
    for cat, total in totals.items():
        print(f"{cat}: Rs {total:.2f}")
    print(f"Total: Rs {grand_total:.2f}")
```

```python
def main():
    """Main program control loop."""
    expenses = load_expenses()

    try:
        while True:
            print("\n=== Expense Tracker ===")
            print("1. Add expense")
            print("2. View all")
            print("3. View totals by category")
            print("4. Save and quit")

            choice = input("Choice: ").strip()

            if choice == "1":
                add_expense(expenses)
            elif choice == "2":
                view_all_expenses(expenses)
            elif choice == "3":
                view_totals_by_category(expenses)
            elif choice == "4":
                save_expenses(expenses)
                print("Goodbye!")
                break
            else:
                print("Invalid choice. Please enter 1, 2, 3, or 4.")
    except (KeyboardInterrupt, EOFError):
        print("\n\nProgram interrupted by user. Exiting cleanly...")
```

```python
# Run the application
main()
```
